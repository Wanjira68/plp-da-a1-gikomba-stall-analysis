```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

print("TASK 1: BASIC SALES ANALYSIS")
print("---------------------------")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f}")
print(f"Lowest sales: KES {lowest_sales:,.2f}")

print(f"Best day: {days[highest_index]}, Week {highest_index // 7 + 1}")
print(f"Worst day: {days[lowest_index]}, Week {lowest_index // 7 + 1}")
```

```python
def week_totals(sales):
    """Calculate total sales for each of the two weeks."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth = ((week2_total - week1_total) / week1_total) * 100

print("TASK 2: WEEKLY SALES COMPARISON")
print("--------------------------------")
print(f"Week 1 total: KES {week1_total:,.2f}")
print(f"Week 2 total: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth:.2f}%")
```

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal, or Weak."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


categories = {"Strong": 0, "Normal": 0, "Weak": 0}

print("TASK 3: DAILY SALES CLASSIFICATION")
print("-----------------------------------")

for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    categories[category] += 1
    print(f"{day}: KES {amount:,.2f} - {category}")

print("\nCategory totals:")
print(categories)
```

```python
weekday_sales = daily_sales[:5] + daily_sales[7:12]
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

difference = ((weekend_average - weekday_average) / weekday_average) * 100

print("TASK 4: WEEKDAY VS WEEKEND SALES")
print("---------------------------------")
print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
print(f"Weekend sales difference: {difference:.2f}%")
```

```python
def products_to_reorder(stock):
    """Return products whose quantity is at or below reorder level."""
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


print("TASK 5: STOCK ANALYSIS")
print("-----------------------")

print("Products to reorder:")
print(products_to_reorder(stock))

print("\nGross profit margins:")

for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")
```

```python
print("=" * 40)
print("       STALL PERFORMANCE REPORT")
print("=" * 40)

print(f"Total sales (14 days): KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")

print(f"\nWeek 1 sales: KES {week1_total:,.2f}")
print(f"Week 2 sales: KES {week2_total:,.2f}")
print(f"Sales growth: {growth:.2f}%")

print(f"\nHighest daily sales: KES {highest_sales:,.2f}")
print(f"Lowest daily sales: KES {lowest_sales:,.2f}")

print(f"\nAverage weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")

print("\nProducts to reorder:")
for product in products_to_reorder(stock):
    print(f"- {product}")

print("\nGross profit margins:")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"- {product}: {margin:.2f}%")

print("=" * 40)
print("             END OF REPORT")
print("=" * 40)
```
